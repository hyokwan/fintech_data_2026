### 1. 데이터 행 조회하기

```python
customerData.head(2)
```

```python
# 조건은 데이터 타입을 정확히 맞춰 줘야한다
targetEmi = "3"
# 무이자할부가 3건 이면서 평균구매가격이 3000달러 이상인 고객을 조회해줘
resultData = customerData.loc[ (customerData.EMI == targetEmi)]
resultData
```

```python
# 다중 조건은 각 조건을 괄호로 묶어줘야 한다.
targetEmi = 3
targetPrice = 4000
# 무이자할부가 3건 이면서 평균구매가격이 3000달러 이상인 고객을 조회해줘
resultData = customerData.loc[ (customerData.EMI == targetEmi) &
            (customerData.AVGPRICE >= targetPrice)     ]
resultData
```

```python
## 특정 컬럼에 대한 다중 값 조회 시 isin
## 해당 부정 표현 NOT IN isin 조건 앞에 ~ 를 붙여주면 된다.
targetEmiList = [1,3]

# 무이자할부가 3건 이면서 평균구매가격이 3000달러 이상인 고객을 조회해줘
resultData = customerData.loc[ ~customerData.EMI.isin(targetEmiList) ]
resultData
```
