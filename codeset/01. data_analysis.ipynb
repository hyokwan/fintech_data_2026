```python
### 데이터 조작 pandas
### 사용하는 방법이 람다~
### 1. np.where: 간단한 데이터 처리 시
### 2. loc: 딱 수정포인트만 조건으로 검색해서 수정하고 빠지기
### 3. function: 복잡한 로직인 경우 함수를 정의해서 처리
### 4. dictionary 활용: 딕셔너리 정의 후 매핑: 문자 -> 숫자로 변환
### 5. labelEncoder 활용: 카테고리 데이터를 숫자로 매핑 (근데 카테고리가 많은 경우)
### 6. onehotEncoder 활용: 인공지능에서 이미지 학습   1 -> 00001, 2-> 00010, 3 -> 00100
```

```python
avgPriceMean = customerData.AVGPRICE.mean()
avgPriceMean
```

```python
%%timeit
### np.where 활용방법: 데이터프레임명["컬럼명"] = np.where( 조건식, True, False)
customerData["NEW_CAT"] = np.where(customerData.AVGPRICE < avgPriceMean, 0, 1 )
# customerData.loc[ (customerData.NEW_CAT < 1) & (customerData.AVGPRICE >= avgPriceMean) ]
```

```python
%%timeit
### loc 기능 (평균가격보다 작게 구매금액이 형성되어 있으면 -> 0 )
### loc 속도가 좀 np.where보다 떨어진다
customerData.loc[ customerData.AVGPRICE < avgPriceMean, "NEW_CAT2" ] = 0
customerData.loc[ customerData.AVGPRICE >= avgPriceMean, "NEW_CAT2" ] = 1
```

```python
# customerData.loc[ customerData.NEW_CAT != customerData.NEW_CAT2 ]
```

```python
def custCategory(inValue):
    ### avgPriceMean : 고객들의 평균 구매가격
    # 각 고객들의 AVGPRICE 값
    # inValue = 2000
    # 리턴 Value
    outValue = 0
    # 각 고객들의 AVGPRICE 가 전체 고객의 평균 AVGPRICE 보다 작으면 0 아니면 1 
    if inValue < avgPriceMean:
        outValue = 0
    else:
        outValue = 1
    return outValue
```

```python
customerData["NEW_CAT3"] = customerData.AVGPRICE.apply(custCategory)
```

### 6. 이동평균

```python
kopoData = pd.read_csv("../dataset/kopo_product_volume.csv")
```

```python
# ST0001 데이터만 선택
targetProduct = "ST0001"
st0001Data = kopoData.loc[ kopoData.PRODUCTGROUP == targetProduct]
# 데이터를 정렬한다 (기본키): null 가질수 없고 데이터에서 중복을 허락하지 않는 컬럼
# 기본키: REGIONID, PRODUCTGROUP, YEARWEEK
# 파밀리데이터 기본키 (데이터베이스를 SQL)
```

```python
### 중복을 허락하지 않는거는 REGIONID, PRODUCTGROUP, YEARWEEK는 중복이
groupKey = ["REGIONID","PRODUCTGROUP","YEARWEEK"]
testData = st0001Data.groupby(by=groupKey)["VOLUME"].agg(["count"]).reset_index()

testData = testData.rename(columns = {"count":"CNT"})
testData.loc[ testData.CNT != 1 ]
# 삼성SDS 후임.. 1년도.. 단한개의 코드도... 운영계에다 이관을 해본적이 없다!!
```

```python
sortKey = ["REGIONID","PRODUCTGROUP","YEARWEEK"]
### 이동평균 13구간 (왜: 분기별 평균판매량을 13*4 = 52 )
sortedData = st0001Data.sort_values( by = sortKey ).reset_index(drop=True)
sortedData["MA13"] = \
        sortedData.VOLUME.rolling(window=13, center=True, min_periods=1).mean()
# 이동표준편차 5 구간 (최근 한달 기준 + - 변화데이터셋 구축한거다)
sortedData["STD5"] = \
        sortedData.VOLUME.rolling(window=5, center=True, min_periods=1).std()
# 상한선, 하한선 설정
sortedData["UPPER_BAND"] = sortedData.MA13 + sortedData.STD5
sortedData["LOWER_BAND"] = sortedData.MA13 - sortedData.STD5
# 정제된 판매량값
sortedData["REFINED_VOLUME"] = \
np.where(sortedData.MA13 > sortedData.UPPER_BAND, sortedData.UPPER_BAND,
         np.where(sortedData.MA13 < sortedData.LOWER_BAND, sortedData.LOWER_BAND,sortedData.VOLUME ))
sortedData["MA5"] = sortedData.REFINED_VOLUME.rolling(window=5, center=True, min_periods=1).mean()
## 정제된판매량 / 이동평균선 5구간 (한달간의 평균 대비)
sortedData["SEASONALITY"] = np.where( sortedData["MA5"] == 0 , 1, sortedData.REFINED_VOLUME / sortedData.MA5 )
sortedData
### 지역 상품 연주차별 계절성 지수 산출하기 
sortedData["YEAR"] = sortedData.YEARWEEK.astype(str).str[:4]
sortedData["WEEK"] = sortedData.YEARWEEK.astype(str).str[4:]
### 최종 그룹바이
groupKey = ["REGIONID","PRODUCTGROUP","WEEK"]
seasonResult = sortedData.groupby(by=groupKey)["SEASONALITY"].agg(["mean"]).reset_index()
seasonResult = seasonResult.rename(columns = {"mean":"SEASON_MEAN"})
seasonResult
```

### 7. 데이터 조인

```python
joinMaster = pd.read_csv("../dataset/join_region_master.csv", encoding="ms949")
joinSellout = pd.read_csv("../dataset/join_kopo_product_volume.csv")
```

```python
# STEP1: 샘플 데이터 생성하기
extraValue = ["A05","ST0005",202520,6000]
# STEP2: 샘플데이터를 데이터프레임 형태로 변환하기
extraData = pd.DataFrame( [ extraValue ] )
extraData.columns = joinSellout.columns
extraData
# STEP3: 기존 joinSellout 데이터와 병합하기 (UNION)
# pd.concat( [  데이터프레임#1, 데이터프레임#2  ])
baseDataUnion = pd.concat( [ joinSellout, extraData ] , axis=0   )
baseDataUnion = baseDataUnion.reset_index(drop=True)
baseDataUnion
```

```python
### 데이터 조인
joinKey = ["REGIONID"]
### 데이터 조인 (inner, left, right, outer (full_outer) )
joinData = pd.merge( left= baseDataUnion, right = joinMaster, on=joinKey, how="outer")

### 영상_포트폴리오
## 기술적인 트리
## 이동평균 데이터프레임명[타겟컬럼].rolling(window=구간, center=T/F, min_periods=1 )
## UNION = pd.concat( [데이터프레임#1, 데이터프레임#2], axis=0 )
## JOIN = pd.merge(left, right, on, how)
```
